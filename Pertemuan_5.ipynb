```python
# Soal 1
angka = int(input("Masukkan bilangan bulat antara 1 sampai 5: "))

if angka == 1:
    faktorial = 1
elif angka == 2:
    faktorial = 1 * 2
elif angka == 3:
    faktorial = 1 * 2 * 3
elif angka == 4:
    faktorial = 1 * 2 * 3 * 4
elif angka == 5:
    faktorial = 1 * 2 * 3 * 4 * 5
else:
    print("Input tidak valid. Masukkan bilangan antara 1 sampai 5.")
    faktorial = None

if faktorial is not None:
    print("Faktorial dari", angka, "adalah", faktorial)
```

```python
# Soal 2

# Program Klasifikasi Kualitas Daya

# Meminta input dari pengguna
faktor_daya = float(input("Masukkan faktor daya (0.0 - 1.0): "))
efisiensi = float(input("Masukkan efisiensi distribusi (%): "))

# Validasi input
if faktor_daya < 0.0 or faktor_daya > 1.0:
    print("Input faktor daya tidak valid!")
elif efisiensi < 0 or efisiensi > 100:
    print("Input efisiensi distribusi tidak valid!")
else:
    # Menghitung skor kualitas daya
    skor = faktor_daya * efisiensi

    print("Skor kualitas daya:", skor)

    # Menentukan klasifikasi kualitas daya
    if skor >= 85 and faktor_daya >= 0.90:
        print("Klasifikasi: PRIMA — sistem dalam kondisi optimal")
    elif skor >= 70 and faktor_daya >= 0.80:
        print("Klasifikasi: BAIK — sistem memadai, pantau rutin")
    elif skor >= 50 or faktor_daya >= 0.75:
        print("Klasifikasi: CUKUP — ada ruang perbaikan")
    else:
        print("Klasifikasi: BURUK — perlu perbaikan segera")
```

```python
# Soal 3

# 1. Meminta input dari pengguna
iradiasi = float(input("Masukkan iradiasi matahari (W/m2): "))
SoC = float(input("Masukkan SoC baterai (%): "))

# 2. Validasi input
if iradiasi < 0:
    print("Input iradiasi tidak valid!")
elif SoC < 0 or SoC > 100:
    print("Input SoC tidak valid! Masukkan nilai 0 sampai 100.")
else:
    # 3. Menghitung estimasi daya panel
    daya_panel = iradiasi * 0.18 * 10

    print("Estimasi daya panel:", daya_panel, "W")

    # 4. Menentukan mode operasi
    if iradiasi > 600 and SoC < 80:
        print("Mode Operasi: CHARGING")
        print("Iradiasi tinggi, baterai belum penuh, prioritaskan pengisian.")

    elif iradiasi > 600 and SoC >= 80:
        print("Mode Operasi: GRID-EXPORT")
        print("Baterai sudah penuh, ekspor kelebihan daya ke jaringan.")

    elif iradiasi > 200 and SoC >= 20:
        print("Mode Operasi: NORMAL")
        print("Iradiasi dan baterai cukup, operasi mandiri.")

    elif iradiasi <= 200 and SoC >= 50:
        print("Mode Operasi: BATTERY-ONLY")
        print("Iradiasi rendah, andalkan baterai.")

    else:
        print("Mode Operasi: EMERGENCY")
        print("Iradiasi rendah dan baterai hampir habis, kurangi beban.")
```
