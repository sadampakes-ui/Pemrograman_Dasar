```python
# Soal 1 week 4

jumlah_buku = int(input("Masukkan jumlah buku: "))
harga_per_buku = int(input("Masukkan harga per buku: Rp "))
diskon_persen = float(input("Masukkan diskon (%): "))
uang_budi = int(input("Masukkan uang Budi: Rp "))
total_belanja = jumlah_buku * harga_per_buku
besar_diskon = total_belanja * diskon_persen / 100
jumlah_dibayar = total_belanja - besar_diskon
kembalian = uang_budi - jumlah_dibayar
iuran_per_orang = int(jumlah_dibayar // 4)
sisa = int(jumlah_dibayar % 4)
uang_cukup = uang_budi >= jumlah_dibayar


print("\n=== HASIL PERHITUNGAN ===")
print("Total belanja : Rp", int(total_belanja))
print("Besar diskon : Rp", int(besar_diskon))
print("Jumlah yang dibayar: Rp", int(jumlah_dibayar))
print("Uang kembalian : Rp", int(kembalian))
print("Iuran per orang : Rp", iuran_per_orang)
print("Sisa pembagian : Rp", sisa)
print("Apakah uang cukup? :", uang_cukup)
```

```python
# Soal 2 Week 4

jumlah_panel = int(input("Masukkan jumlah panel surya: "))
daya_panel = float(input("Masukkan daya per panel (Wp): "))
penyinaran = float(input("Masukkan penyinaran efektif (jam/hari): "))
efisiensi = float(input("Masukkan efisiensi sistem (desimal): "))
kebutuhan = float(input("Masukkan kebutuhan listrik (kWh/hari): "))
tarif = float(input("Masukkan tarif listrik (Rp/kWh): "))
daya_total_wp = jumlah_panel * daya_panel
daya_total_kw = daya_total_wp / 1000
energi_harian = daya_total_kw * penyinaran * efisiensi
energi_bulanan = energi_harian * 3
nilai_listrik = energi_bulanan * tarif
kekurangan = kebutuhan - energi_harian
mencukupi = energi_harian >= kebutuhan


print("\n=== HASIL PERHITUNGAN PLTS ===")
print("Daya total panel :", daya_total_wp, "Wp")
print("Energi per hari :", energi_harian, "kWh")
print("Energi per bulan :", energi_bulanan, "kWh")
print("Nilai listrik per bulan: Rp", nilai_listrik)
print("Kekurangan energi/hari :", kekurangan, "kWh")
print("PLTS mencukupi? :", mencukupi)
```

```python
# Latihan 3 Week 4

daya_generator = float(input("Masukkan daya generator (kW): "))
jam_operasi = float(input("Masukkan waktu operasi per hari (jam): "))
konsumsi_bahan_bakar = float(input("Masukkan konsumsi bahan bakar (liter/kWh): "))
harga_solar = float(input("Masukkan harga solar (Rp/liter): "))
kapasitas_tangki = float(input("Masukkan kapasitas tangki (liter): "))
arus = float(input("Masukkan arus kabel (A): "))
hambatan = float(input("Masukkan hambatan kabel (Ohm): "))
energi_harian = daya_generator * jam_operasi
solar_harian = energi_harian * konsumsi_bahan_bakar
biaya_solar = solar_harian * harga_solar
hari_penuh = int(kapasitas_tangki // solar_harian)
sisa_solar = kapasitas_tangki % solar_harian
rugi_daya = arus ** 2 * hambatan
rugi_tinggi = (rugi_daya > 150) and (arus >= 20)


print("\n=== HASIL PERHITUNGAN ===")
print("Energi per hari :", energi_harian, "kWh")
print("Solar per hari :", solar_harian, "liter")
print("Biaya solar per hari : Rp", biaya_solar)
print("Tangki cukup :", hari_penuh, "hari penuh")
print("Sisa solar :", sisa_solar, "liter")
print("Rugi daya kabel :", rugi_daya, "W")
print("Rugi kabel tergolong tinggi?", rugi_tinggi)
```
