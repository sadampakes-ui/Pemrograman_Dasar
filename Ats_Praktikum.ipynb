```python
# SOAL NO.2

daya_kw = 450
jam = 8
energi_aktual = 7200
kapasitas_kw = 400
energi = daya_kw * jam
capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100


print(f"Energi harian PLTD = {energi} kWh")
print(f"Capacity Factor = {capacity_factor:.2f}%")


#Cara Kerja Code
#1. daya_kw = 450 menyimpan nilai daya pembangkit.
#2. jam = 8 menyimpan lama pembangkit beroperasi.
#3. energi_aktual = 7200 menyimpan energi aktual yang dihasilkan.
#4. kapasitas_kw = 400 menyimpan kapasitas pembangkit.
#5. energi = daya_kw * jam menghitung energi berdasarkan daya dikalikan waktu.
#6. capacity_factor menghitung persentase pemanfaatan kapasitas pembangkit.
#7. print() menampilkan hasil perhitungan.
#8. :.2f membatasi angka desimal menjadi dua angka di belakang koma.
#Hasil energi harian = 3.600 kWh dan capacity factor = 75,00%.
```

```python
# SOAL NO.6

# Input suhu dan tekanan boiler
suhu = float(input("Masukkan suhu uap (°C): "))
tekanan = float(input("Masukkan tekanan uap (bar): "))

# Menentukan status boiler
if suhu >= 550 or tekanan >= 14:
    print("STATUS: TRIP")

elif suhu > 520 and tekanan > 12:
    print("STATUS: WARNING")

else:
    print("STATUS: NORMAL")

#Cara Kerja Code
#1. input() meminta pengguna memasukkan suhu dan tekanan.
#2. float() mengubah input menjadi angka desimal.
#3. if memeriksa apakah suhu mencapai 550°C atau tekanan mencapai 14 bar.
#4. or berarti cukup salah satu kondisi terpenuhi untuk menghasilkan TRIP.
#5. elif memeriksa apakah suhu lebih dari 520°C dan tekanan lebih dari 12 bar.
#6. and berarti kedua kondisi harus terpenuhi untuk menghasilkan WARNING.
#7. else menghasilkan NORMAL jika kedua kondisi sebelumnya tidak terpenuhi.
#Hasil status boiler dapat berupa TRIP, WARNING, atau NORMAL.
#Mengapa TRIP diperiksa lebih dahulu?Karena TRIP merupakan kondisi yang lebih kritis. Jika kondisi TRIP dan WARNING sama-sama terpenuhi, program harus menghasilkan TRIP.
```

```python
# SOAL NO.8

volume = 100
while volume > 0:
    volume -= 20

    print(f"Volume : {volume} liter")
    if volume == 20:
      print("WARNING: Hampir kosong!")
print("TANKI KOSONG - Pengosongan berhenti.")

#Cara Kerja Code
#1. volume = 100 menentukan volume awal tangki.
#2. while volume > 0 membuat perulangan selama volume masih lebih dari nol.
#3. volume -= 20 mengurangi volume sebanyak 20 liter setiap perulangan.
#4. print() menampilkan volume tangki setelah pengurangan.
#5. if volume == 20 memeriksa apakah volume sudah mencapai 20 liter.
#6. Jika kondisi benar, program menampilkan peringatan hampir kosong.
#7. Ketika volume mencapai nol, kondisi while menjadi salah sehingga perulangan berhenti.
#8. Program kemudian menampilkan pesan bahwa tangki sudah kosong.
#Hasil Volume tangki berkurang dari 100, 80, 60, 40, 20, hingga 0 liter.
```

```python
# SOAL NO.10

data = ["62.5", "58.0", "81.3", "N/A", "66.0"]
jumlah_valid = 0

for i, nilai in enumerate(data, start=1):
    try:
        suhu = float(nilai)
        jumlah_valid += 1

        if suhu > 70.0:
            print(f"Warning: Titik {i} suhu bearing tinggi!")

    except ValueError:
        print(f"Titik {i}: data tidak valid - pemeriksaan dihentikan")
        break

print(f"Jumlah data valid = {jumlah_valid}")

#Cara Kerja Code
#1. data = [...] menyimpan lima data suhu bearing dalam bentuk teks.
#2. jumlah_valid = 0 digunakan sebagai penghitung data yang berhasil diperiksa.
#3. for digunakan untuk membaca setiap data dalam list.
#4. enumerate(data, start=1) memberikan nomor titik mulai dari angka 1.
#5. try mencoba mengubah data teks menjadi angka menggunakan float().
#6. jumlah_valid += 1 menambah penghitung apabila data berhasil dikonversi.
#7. if suhu > 70.0 memeriksa apakah suhu melebihi batas 70°C.
#8. except ValueError menangani kesalahan ketika data tidak bisa diubah menjadi angka.
#9. break menghentikan perulangan ketika ditemukan data tidak valid.
#10. print() terakhir menampilkan jumlah data valid yang berhasil diperiksa.
#Hasil terdapat 3 data valid, dan titik ke-3 mendapatkan peringatan suhu tinggi.
#Hal penting: Data titik ke-5 tidak diperiksa karena perulangan sudah dihentikan ketika menemukan "N/A" pada titik ke-4.
```

```python
# CHALLENGE

# Program Pemantauan Konsumsi Energi Listrik Harian
# Program digunakan untuk memantau penggunaan energi selama 7 hari.

energi_harian = []
for hari in range(1, 8):
    try:
        energi = float(input(f"Masukkan konsumsi energi hari ke-{hari} (kWh): "))

        if energi < 0:
            print("Data tidak boleh negatif!")
            continue

        energi_harian.append(energi)

    except ValueError:
        print("Input tidak valid! Data hari ini dilewati.")
print("\n=== HASIL PEMANTAUAN ENERGI ===")

total_energi = 0

for i, energi in enumerate(energi_harian, start=1):
    total_energi += energi

    if energi > 500:
        status = "TINGGI"
    elif energi >= 300:
        status = "NORMAL"
    else:
        status = "RENDAH"

    print(f"Hari ke-{i}: {energi:.2f} kWh - {status}")

if len(energi_harian) > 0:
    rata_rata = total_energi / len(energi_harian)

    print("\nTotal energi =", total_energi, "kWh")
    print(f"Rata-rata energi = {rata_rata:.2f} kWh")

    if rata_rata > 500:
        print("STATUS: Konsumsi energi sangat tinggi!")
    elif rata_rata >= 300:
        print("STATUS: Konsumsi energi normal.")
    else:
        print("STATUS: Konsumsi energi rendah.")

else:
    print("Tidak ada data energi yang valid.")

#Cara Kerja Program
#1. Program meminta pengguna memasukkan data konsumsi energi listrik selama 7 hari.
#2. Data yang dimasukkan diubah menjadi tipe float.
#3. Jika input tidak valid, program menampilkan pesan kesalahan dan melewati data tersebut.
#4. Data yang valid disimpan ke dalam list energi_harian.
#5. Program memeriksa setiap data menggunakan perulangan for.
#6. Konsumsi energi lebih dari 500 kWh dikategorikan tinggi, 300–500 kWh normal, dan di bawah 300 kWh rendah.
#7. Program menghitung total serta rata-rata konsumsi energi selama periode pemantauan.
```
