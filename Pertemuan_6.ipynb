```python
# Soal 6F

scores = [80, 65, 90, 75, 70]

print("range(5) gives:", list(range(5)))

# 1) for-each: mengambil setiap nilai
for score in scores:
    print("score:", score)

# 2) by position: mengambil nilai berdasarkan index
for i in range(len(scores)):
    print("index", i, "holds", scores[i])

# 3) Menghitung total skor
total = 0
for score in scores:
    total += score

print("Total score:", total)

# 4) Menghitung rata-rata
average = total / len(scores)
print("Average score:", average)

# 5) Menghitung skor >= 75
count = 0
for score in scores:
    if score >= 75:
        count += 1

print("Scores >= 75:", count)
```

```python
# Soal 6G Program before edit

temps = [28, 31, 30, 33, 29, 27, 32]

# Daily change: start at index 1 because we look at i - 1
for i in range(1, len(temps)):
    change = temps[i] - temps[i - 1]
    print(f"day {i + 1}: {temps[i]} °C, change {change:+d}")

# 3-day moving average: window = the 3 values ending at index i
avg3 = []
for i in range(2, len(temps)):
    window = temps[i - 2 : i + 1]
    avg3.append(round(sum(window) / 3, 1))

print("3-day averages:", avg3)
print(len(temps), "values ->", len(avg3), "averages")
```

```python
# 6G Program After Edit

temps = [28, 31, 30, 33, 29, 27, 32]

# Daily change
for i in range(1, len(temps)):
    change = temps[i] - temps[i - 1]
    print(f"day {i + 1}: {temps[i]} °C, change {change:+d}")

# 5-day moving average
avg5 = []
for i in range(4, len(temps)):
    window = temps[i - 4 : i + 1]
    avg5.append(round(sum(window) / 5, 1))

print("5-day averages:", avg5)
print(len(temps), "values ->", len(avg5), "averages")
```

```python
# Soal 6H

energy = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]

# Daily change: start at index 1 because each value is compared with the one before
for i in range(1, len(energy)):
    change = (energy[i] - energy[i - 1]) / energy[i - 1] * 100
    if change < -20:
        print(f"Day {i + 1}: {change:+.1f}% versus the day before, investigate")

# 7-day moving average
avg_7 = []
for i in range(6, len(energy)):
    window = energy[i - 6 : i + 1]
    avg_7.append(sum(window) / 7)

print(f"{len(avg_7)} moving-average values ({len(energy)} days of data)")
print(f"Lowest {min(avg_7):,.0f} MWh/day, highest {max(avg_7):,.0f} MWh/day")
```

```python
# Soal 6J

marks = [
    [80, 70, 90],     # row 0: Ana
    [65, 75, 60],     # row 1: Budi
    [90, 85, 95],     # row 2: Cici
]

print(marks[1])        # a whole row
print(marks[1][2])     # row 1, column 2: Budi's test 3
print("rows:", len(marks), " columns:", len(marks[0]))

# A total for each row (each student)
for r in range(len(marks)):
    print("Student", r + 1, "total:", sum(marks[r]))

# A column: the same index in every row (everybody's test 1)
test1 = []
for row in marks:
    test1.append(row[0])
print("Test 1 scores:", test1)
```
